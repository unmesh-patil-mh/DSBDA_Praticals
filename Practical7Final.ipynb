```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Sample Text
text = "Natural language processing is easy and interesting"
```

```python
word = word_tokenize(text)
print("Tokenization: ")
word
```

```python
# Stop words Removal 
stop = stopwords.words("english")
```

```python
new_word = []
for i in word:
    if i not in stop:
        new_word.append(i)

print("After Removal of stopwords:")
print(new_word)
```

```python
# POS Tagging
print("POS Tagging: ")
print(nltk.pos_tag(word))
```

```python
# Stemming
from nltk.stem import PorterStemmer
ps = PorterStemmer()
for i in new_word:
    print(ps.stem(i))
```

```python
# Lemminization
from nltk.stem import WordNetLemmatizer
lm = WordNetLemmatizer()

for i in new_word:
    print(lm.lemmatize(i))
```
