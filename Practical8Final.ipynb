```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = sns.load_dataset("titanic")
```

```python
data
```

```python
sns.countplot(x="survived", data=data)
plt.title("Survived Passengers")
plt.show()
```

```python
sns.countplot(x="sex", data=data)
plt.title("Passenger Gender")
plt.show()
```
